### Question 6

Using the `starwars` dataset, identify the individuals with:

- the tallest height
- the lowest mass

HINT: the starwars dataset is part of the dplyr package

```r
library(tidyverse) #load tidyverse to access dplyr
#find the tallest height 
tallest <- starwars %>%
    filter(height == max(height, na.rm = TRUE)) %>%
    select(name, height, species, mass)
print(tallest)    
# find the lowest mass
lowest <- starwars %>%
    filter(mass == min(mass, na.rm = TRUE)) %>%
    select(name, height, species, mass)
print(lowest)
```

### Question 7

The `dplyr` dataset `storms` is part of the US National Oceanic and Atmospheric Administration database of Atlantic hurricanes. It contains multiple repeated measurements from hurricanes. For a given storm, there are measurements from every 6 hours across the life time of the storm.

Use a tidyverse approach with `dplyr` to answer the following:

- how many different storms are contained in the database?
- which year had the most storms?
- which storm, in which year had the highest pressure?

HINT:
unique() will help ensure you remove duplicated data from a vector e.g.
x <- c(1, 1, 2, 2, 3, 3, 4, 4)
unique(x)

```r
library(tidyverse)#load tidyverse to access dplyr
# to find how many different storms are contained in the database
storms_name <- storms %>%
    distinct(name) %>%
    tally()

unique(storms_name)# Total 260

# find which year had the most
most_storms <- storms %>%
    group_by(year) %>%
    summarise(storm_n = n_distinct(name)) %>%
    arrange(desc(storm_n)) %>%
    slice(1)
    
print(most_storms)#2020years have the most stroms,30 storms

# find the highest pressure storms
highest_pressure <- storms %>%
    arrange(desc(pressure)) %>%
    select(name, year, pressure) %>%
    slice(1)

print(highest_pressure)
```

### Question 8

The `ggplot2` dataset `diamonds` contains the price and information from the sale of 54,000 different diamonds. You can use ?diamonds to find out more about the dataset if you want.

Using `ggplot2`:

- plot the relationship between the carat (i.e. the weight of the diamond) and the price.
- plot a histogram of prices of all the diamonds
- plot a boxplot of the difference in price between the different cuts

```r
install.packages("ggplot2") # Install ggplot2 package
# Question8.1
ggplot(aes(x = carat, y = price), data = diamonds) +#Using diamonds data and map carat to x-axis and price to the y-axis
    geom_point() + # add point
    labs(
        x = "Carat",
        y = "Price",
        title = "The relationship between the carat and the price"
    )#done
#Question 8.2
ggplot(aes(x = price), data = diamonds)+ 
  geom_histogram() +
  labs(
      x = "Price",
      y = "Count",
      title = "Price of all the diamonds"
  )
#Question 8.3
ggplot(aes(x = cut, y = price), data = diamonds) +#Using diamonds data and map cut to x-axis and price to the y-axis
    geom_boxplot() + # add boxplot
    labs(
        x = "Cut",
        y = "Price",
        title = "The difference in price between the different cuts"
    )#done
```

### Question 10

Sticking with the `diamonds` dataset for this question. A carat is a special unit of weight for gemstones where a single carat is 200 mg or 0.2 g. Write a function to convert the `diamonds$carat` vector into grams or milligrams. How heavy is the heaviest diamond?

```r
library(tidyverse)
grams <- function(carat_to_grams){ #convert function from carat to gram
    carat_to_grams * 0.2
}
diamonds$gram <- grams(diamonds$carat)

the_heaviest_diamond <- max(diamonds$gram)
the_heaviest_diamond
```
